```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Tool': ['gTTS', 'MetaMMS', 'IndicTTS', 'GoogleCloudTTS', 'BhashiniIITM'],
    'WSM_Score': [0.81, 0.11, 0.04,  0.92, 0.26],  # avg WSM mean scores of three scenarios
    'CMOS_Mean': [2.45, 0.14, 0.62, 2.79, 1.69],       # CMOS mean scores
    'CMOS_STD': [0.50, 1.48, 1.15, 0.41, 0.54]       # CMOS standard deviation
}

df = pd.DataFrame(data)

corr_matrix = df[['WSM_Score', 'CMOS_Mean', 'CMOS_STD']].corr(method='pearson')
print("Correlation Matrix:\n", corr_matrix)

plt.figure(figsize=(6,5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix between WSM and CMOS Metrics")
plt.show()
```
